```python
inventario = [
    {
        "codigo": "L001",
        "titulo": "Python Básico",
        "autor": "Juan Pérez",
        "cantidad": 10,
        "precio": 19990
    },
    {
        "codigo": "L002",
        "titulo": "Data Science con R",
        "autor": "María López",
        "cantidad": 5,
        "precio": 25000
    }
]

def obtener_entero(mensaje, minimo=None):
    """Función auxiliar para obtener input entero con validación."""
    while True:
        try:
            valor = int(input(mensaje))
            if minimo is not None and valor < minimo:
                print(f"El valor debe ser al menos {minimo}.")
            else:
                return valor
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número entero.")

def obtener_flotante(mensaje, minimo=None):
    """Función auxiliar para obtener input flotante con validación."""
    while True:
        try:
            valor = float(input(mensaje))
            if minimo is not None and valor < minimo:
                print(f"El valor debe ser al menos {minimo}.")
            else:
                return valor
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número decimal o entero.")


def registrar_libro():
    """Registra un nuevo libro en el inventario."""
    print("\n--- Registrar Libro ---")
    while True:
        codigo = input("Ingrese código del libro (ej. L003): ").strip().upper()
        if not codigo:
            print("El código no puede estar vacío.")
            continue
        if any(libro['codigo'] == codigo for libro in inventario):
            print(f"Error: El código '{codigo}' ya existe. Ingrese uno diferente.")
        else:
            break

    titulo = input("Ingrese título del libro: ").strip()
    if not titulo:
        print("El título no puede estar vacío.")
        return

    autor = input("Ingrese autor del libro: ").strip()
    if not autor:
        print("El autor no puede estar vacío.")
        return

    cantidad = obtener_entero("Ingrese cantidad disponible: ", minimo=0)
    precio = obtener_flotante("Ingrese precio: ", minimo=0.0)

    nuevo_libro = {
        "codigo": codigo,
        "titulo": titulo,
        "autor": autor,
        "cantidad": cantidad,
        "precio": precio
    }
    inventario.append(nuevo_libro)
    print(f"Libro '{titulo}' registrado exitosamente.")


def buscar_libro():
    """Busca un libro por su código y muestra su información."""
    print("\n--- Buscar Libro ---")
    codigo = input("Ingrese el código del libro a buscar: ").strip().upper()

    libro_encontrado = None
    for libro in inventario:
        if libro['codigo'] == codigo:
            libro_encontrado = libro
            break

    if libro_encontrado:
        print("\n--- Información del Libro ---")
        print(f"Código: {libro_encontrado['codigo']}")
        print(f"Título: {libro_encontrado['titulo']}")
        print(f"Autor: {libro_encontrado['autor']}")
        print(f"Cantidad: {libro_encontrado['cantidad']}")
        print(f"Precio: ${libro_encontrado['precio']:,}")
        print("________________________")
    else:
        print(f"Error: Libro con código '{codigo}' no encontrado.")


def actualizar_stock():
    """Actualiza la cantidad disponible de un libro."""
    print("\n--- Actualizar Stock ---")
    codigo = input("Ingrese el código del libro a actualizar stock: ").strip().upper()

    libro_encontrado = None
    for libro in inventario:
        if libro['codigo'] == codigo:
            libro_encontrado = libro
            break

    if libro_encontrado:
        print(f"Stock actual de '{libro_encontrado['titulo']}': {libro_encontrado['cantidad']}")
        nueva_cantidad = obtener_entero("Ingrese la nueva cantidad disponible: ", minimo=0)
        libro_encontrado['cantidad'] = nueva_cantidad
        print(f"Stock de '{libro_encontrado['titulo']}' actualizado a {nueva_cantidad} unidades.")
    else:
        print(f"Error: Libro con código '{codigo}' no encontrado.")


def mostrar_inventario():
    """Muestra todos los libros registrados en el inventario."""
    print("\n--- Inventario Completo ---")
    if not inventario:
        print("El inventario está vacío.")
        return

    for libro in inventario:
        print(f"Código: {libro['codigo']}")
        print(f"Título: {libro['titulo']}")
        print(f"Autor: {libro['autor']}")
        print(f"Cantidad: {libro['cantidad']}")
        print(f"Precio: ${libro['precio']:,}")
        print("________________________")


def libro_mas_caro():
    """Identifica y muestra la información del libro más caro."""
    print("\n--- Libro Más Caro ---")
    if not inventario:
        print("El inventario está vacío.")
        return

    libro_caro = inventario[0]
    for libro in inventario:
        if libro['precio'] > libro_caro['precio']:
            libro_caro = libro

    print("\n--- Información del Libro Más Caro ---")
    print(f"Código: {libro_caro['codigo']}")
    print(f"Título: {libro_caro['titulo']}")
    print(f"Autor: {libro_caro['autor']}")
    print(f"Cantidad: {libro_caro['cantidad']}")
    print(f"Precio: ${libro_caro['precio']:,}")
    print("________________________")


def eliminar_libro():
    """Elimina un libro del inventario por su código."""
    print("\n--- Eliminar Libro ---")
    codigo = input("Ingrese el código del libro a eliminar: ").strip().upper()

    libro_encontrado_idx = -1
    for i, libro in enumerate(inventario):
        if libro['codigo'] == codigo:
            libro_encontrado_idx = i
            break

    if libro_encontrado_idx != -1:
        libro_eliminado = inventario.pop(libro_encontrado_idx)
        print(f"Libro '{libro_eliminado['titulo']}' (código: {codigo}) eliminado exitosamente.")
    else:
        print(f"Error: Libro con código '{codigo}' no encontrado.")


def mostrar_menu():
    """Muestra el menú principal de la aplicación."""
    print("\n===== LIBRERÍA =====")
    print("1. Registrar libro")
    print("2. Buscar libro")
    print("3. Actualizar stock")
    print("4. Mostrar inventario")
    print("5. Mostrar libro más caro")
    print("6. Eliminar libro")
    print("7. Salir")
    print("====================")


while True:
    mostrar_menu()
    opcion = obtener_entero("Seleccione una opción: ", minimo=1)

    if opcion == 1:
        registrar_libro()
    elif opcion == 2:
        buscar_libro()
    elif opcion == 3:
        actualizar_stock()
    elif opcion == 4:
        mostrar_inventario()
    elif opcion == 5:
        libro_mas_caro()
    elif opcion == 6:
        eliminar_libro()
    elif opcion == 7:
        print("Saliendo del programa. ¡Hasta pronto!")
        break
    else:
        print("Opción inválida. Por favor, intente de nuevo.")
```
